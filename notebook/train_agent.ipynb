```python
import sys
import os
import gymnasium as gym
from stable_baselines3 import PPO
from stable_baselines3.common.callbacks import EvalCallback

# 1. Get the absolute path to your project root
# (Assuming your notebook is inside a subfolder, like mujoco_project/notebooks)
project_root = os.path.dirname(os.path.abspath(''))
sys.path.append(project_root)

from src.envs import CustomCartPoleEnv

# 2. Build the absolute path to the XML file
absolute_xml_path = os.path.join(project_root, "assets", "cartpole.xml")

# 3. Initialize the environment with the absolute path
env = CustomCartPoleEnv(xml_path=absolute_xml_path)
```

```python
# CELL 2
print("--- WHAT THE AGENT CAN DO ---")
print(f"Action Space: {env.action_space}")
# If it's a continuous CartPole, it might print Box([-1.], [1.]), meaning it can push left or right with varying force.

print("\n--- WHAT THE AGENT SEES ---")
print(f"Observation Space: {env.observation_space}")
# This will show you the bounds of the numbers the agent receives (e.g., cart position, pole angle).

# Let's take exactly ONE step manually, without AI
obs, info = env.reset()
print(f"\nStarting Observation (State): {obs}")

# Pick a completely random action
random_action = env.action_space.sample()
print(f"Taking Random Action: {random_action}")

# Feed the action to the environment
new_obs, reward, terminated, truncated, info = env.step(random_action)

print(f"New Observation: {new_obs}")
print(f"Reward Received: {reward}")
print(f"Did the episode end?: {terminated}")
```

```python
# CELL 3
model = PPO("MlpPolicy", env, verbose=1)

# Let's look at the actual PyTorch Neural Network architecture!
print(model.policy)
```

```python
# CELL 4
# We use a tiny number just to watch the logs appear
model.learn(total_timesteps=5000)
```
